# Module 4: Multilayer Perceptron

Forward propagation of a multilayer perceptron using NumPy.

* Layer 0: 3 input nodes
* Layer 1: 5 nodes (sigmoid)
* Layer 2: 5 nodes (sigmoid)
* Layer 3: 2 nodes (sigmoid)
* Layer 4: 2 output nodes (ReLU)

```python
import numpy as np
from multilayer_perceptron import MultilayerPerceptron
```

## Create the network

```python
mlp = MultilayerPerceptron()

# shapes of the weights and biases
for name, value in mlp.params.items():
    print(name, value.shape)
```

## Forward propagation with one input

```python
x = np.array([1.0, 0.5, -1.0])
y = mlp.forward(x)

print("input:", x)
print("layer 1 (sigmoid):", mlp.h1)
print("layer 2 (sigmoid):", mlp.h2)
print("layer 3 (sigmoid):", mlp.h3)
print("layer 4 (relu):", y)
```

## Forward propagation with more inputs

```python
samples = [
    np.array([0.0, 0.0, 0.0]),
    np.array([-2.0, 1.0, 3.0]),
    np.array([5.0, -3.0, 2.0]),
]

for s in samples:
    print(f"input: {s}, output: {mlp.forward(s)}")
```
